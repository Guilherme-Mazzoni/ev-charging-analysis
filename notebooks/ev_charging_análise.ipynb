À medida que os veículos elétricos (VEs) se tornam mais populares, há uma necessidade crescente de acesso a estações de recarga, também conhecidas como pontos de carregamento. Para isso, muitos prédios residenciais modernos começaram a adaptar suas garagens para incluir estações de recarga compartilhadas. Uma estação de recarga é considerada compartilhada se for acessível por qualquer pessoa do prédio.

<img src="charging_station.jpg" alt="Carregamento de VE" width="500" height="auto">

Mas, com o aumento da demanda, cresce também a competição por esses pontos — nada é mais frustrante do que chegar em casa e não encontrar nenhuma estação de recarga disponível! Neste projeto, você usará um conjunto de dados para ajudar administradores de prédios residenciais a entender melhor os hábitos de recarga de seus moradores.

Os dados foram carregados em um banco de dados PostgreSQL com uma tabela chamada `charging_sessions` com as seguintes colunas:

## charging_sessions

| Coluna | Definição | Tipo de dado |
|-|-|-|
|`garage_id`| Identificador da garagem/prédio |`VARCHAR`|
|`user_id` | Identificador do usuário individual |`VARCHAR`|
|`user_type`| Indica se a estação é `Shared` (Compartilhada) ou `Private` (Privada) | `VARCHAR` |
|`start_plugin`| Data e hora de início da sessão |`DATETIME`|
|`start_plugin_hour`| Hora (em formato 24h) em que a sessão começou | `NUMERIC`|
|`end_plugout`| Data e hora de término da sessão | `DATETIME` |
|`end_plugout_hour`| Hora (em formato 24h) em que a sessão terminou | `NUMERIC`|
|`duration_hours`| Duração da sessão, em horas |`NUMERIC`|
|`el_kwh`| Quantidade de eletricidade utilizada (em quilowatt-hora) |`NUMERIC`|
|`month_plugin`| Mês em que a sessão começou |`VARCHAR`|
|`weekdays_plugin`| Dia da semana em que a sessão começou |`VARCHAR`|

```python
-- unique_users_per_garage
-- Modify the code below
SELECT  garage_id,  
	COUNT(DISTINCT user_id) AS num_unique_users
FROM charging_sessions
WHERE user_type = 'Shared'
GROUP BY garage_id
ORDER BY num_unique_users DESC
```

```python
SELECT weekdays_plugin, start_plugin_hour, COUNT(*) AS num_charging_sessions
FROM charging_sessions
WHERE USER_TYPE = 'Shared'
GROUP BY weekdays_plugin, start_plugin_hour
ORDER BY num_charging_sessions DESC
LIMIT 10 

```

```python
-- long_duration_shared_users
SELECT user_id, AVG(duration_hours) AS avg_charging_duration
FROM charging_sessions
WHERE USER_TYPE = 'Shared' 
GROUP BY user_id
HAVING AVG(duration_hours) > 10
ORDER BY avg_charging_duration DESC
LIMIT 6
```
